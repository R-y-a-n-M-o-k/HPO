```python
import numpy as np
equilibrium_vals = np.array([1.48332676, 1.45443504, 104.3977314])

r_PO = np.array([0,        
                -0.05, 0.05,                       
                -0.10, 0.12,             
                -0.14, 0.18, 0.21,
                ])

r_PO = r_PO + equilibrium_vals[0]

r_PH = np.array([0, 
                -0.05, 0.12,
                -0.16, 0.28,
                -0.23,
                -0.31, 0.72,
                 1.12, 1.36
                ])

r_PH = r_PH + equilibrium_vals[1]

a_HPO = np.array([55, 65, 70, 75, 80, 85])

geometries = []
point = 0

for i in r_PO:
    for j in r_PH:
        for k in a_HPO:
            geometries.append([point, i, j, k])
            point += 1
        
print('Number of Geometries:')
print(len(geometries))
geometries = np.array(geometries)

np.savetxt('HPO_3D_grid_extra.txt', 
           geometries, 
           fmt=('%d','%.6f','%.6f','%.6f'), 
           delimiter=' ')
```
