```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
ab_initio_data = np.loadtxt('1D_energies.txt', delimiter='\t')
min_energy = -416.6777585381
hart_to_cm = 219474.63
#convert hartree to cm^-1
ab_initio_data[:, 3] = (ab_initio_data[:, 3] - min_energy) * hart_to_cm
```

```python
end_idx = [0, 73, 234, 355]
x_coord = [r'$r_1$ (Å)', r'$r_2$ (Å)', r'$\theta$ (°)']
x_lower = [1.23, 1.00, 54]
x_upper = [1.93, 2.6, 170]
colors = ['blue', 'red', 'black']
save_files = ['r1_PES.png', 'r2_PES.png', 'theta_PES.png']

for i in range(3):
    coordinates = ab_initio_data[end_idx[i]:end_idx[i+1], i]
    energies = ab_initio_data[end_idx[i]:end_idx[i+1], 3]

    plt.figure(figsize=(12, 8))
    plt.plot(coordinates, 
             energies, 
             linestyle='-', 
             linewidth = 2.8,
             color=colors[i])
    plt.xlabel(x_coord[i], fontsize=16)
    plt.ylabel(r'Relative Energy $(\text{cm}^{-1})$', fontsize=16)
    plt.xlim(x_lower[i], x_upper[i])

    plt.tick_params(axis='both', labelsize=16)
    plt.savefig(save_files[i], dpi=300, bbox_inches='tight', transparent=True)
```
