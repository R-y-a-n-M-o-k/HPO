```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
colors = ['b-', 'r-', 'k-']
save_files = ['r1_stick.png', 'r2_stick.png', 'theta_stick.png']
x_vals = np.array([1192.863958, 2068.666059, 985.458437])
x_lower = [1015, 1875, 820]
x_upper = [1385, 2255, 1160]

for i in range(3):

    wavenum, intensity = np.loadtxt(f"J0_J141_T300_stick_{i+1}.stick", 
                                    unpack=True)
    plt.figure(figsize=(9, 7), dpi=300)
    markerline, stemlines, baseline = plt.stem(
                                      wavenum, intensity/1e-20,
                                      linefmt=colors[i],
                                      markerfmt=' ',
                                      basefmt='k-',
                                      bottom=1e-30,
                                      #label=f"{i+1}"
    )
    plt.xlim(x_lower[i], x_upper[i])
    plt.ylim(bottom=1e-30*1e20)
    
    plt.setp(stemlines, linewidth=0.15, alpha=1.0)
    plt.setp(baseline, linewidth=0.8)
    plt.tick_params(axis='both', which='major', labelsize=18, 
                    direction='out', length=3)

    plt.ylabel(r'Intensity ($10^{-20}$ cm molecule$^{-1}$)', fontsize=20)
    plt.xlabel(r'Wavenumber (cm$^{-1}$)', fontsize=20)
    plt.savefig(save_files[i], dpi=300, bbox_inches='tight', transparent=True)
    #plt.legend()
```
